```python
from pathlib import Path
import pandas as pd
import numpy as np


import sys

if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.preprocessing import (
    load_clinical_data,
    load_datasets,
    map_patient_subtypes
)

# Diretório principal do projeto
ROOT_DIR = Path.cwd().parent

# Diretórios dos dados
DATA_DIR = ROOT_DIR / "data"
PROCESSED_DIR = DATA_DIR / "processed"
MAD_DIR = PROCESSED_DIR / "mad_datasets"

# Carregar dados clínicos
PATH_CLINICO = PROCESSED_DIR / "clinical_final.csv"

pacientes, dict_subtipos = load_clinical_data(
    PATH_CLINICO
)

print("Pacientes clínicos:", len(dict_subtipos))
print(pacientes["SUBTYPE"].value_counts())

# Carregar cenários de expressão
datasets = load_datasets(MAD_DIR)

# Verificar distribuição dos subtipos em cada cenário
for nome, df in datasets.items():

    y_pam50 = map_patient_subtypes(
        df.index,
        dict_subtipos
    )

    print(f"\n{nome}")
    print(y_pam50.value_counts())
```
